```python
import pandas as pd
import numpy as np

# If the file is .xlsx
file_path = r"C:\Users\harsh\Downloads\online+retail\Online Retail.xlsx"

# Load the dataset
df = pd.read_excel(file_path)

# Verify the data loaded properly
print(df.shape)
df.head()
```

```python
# Drop rows where CustomerID is missing
df_clean = df.dropna(subset=['CustomerID']).copy()

# Convert CustomerID to integer
df_clean['CustomerID'] = df_clean['CustomerID'].astype(int)

# Remove cancelled orders (invoices starting with 'C')
df_clean = df_clean[~df_clean['InvoiceNo'].astype(str).str.startswith('C')]

# Remove non-positive quantities and unit prices
df_clean = df_clean[(df_clean['Quantity'] > 0) & (df_clean['UnitPrice'] > 0)]

# Calculate total monetary value per line item
df_clean['TotalAmount'] = df_clean['Quantity'] * df_clean['UnitPrice']

print("Cleaned rows:", df_clean.shape[0])
print("Unique customers:", df_clean['CustomerID'].nunique())
```

```python
# Set reference date one day after the last recorded transaction
snapshot_date = df_clean['InvoiceDate'].max() + pd.Timedelta(days=1)

# Group by CustomerID
rfm = df_clean.groupby('CustomerID').agg({
    'InvoiceDate': lambda x: (snapshot_date - x.max()).days,
    'InvoiceNo': 'nunique',
    'TotalAmount': 'sum'
}).reset_index()

# Rename columns
rfm.columns = ['CustomerID', 'Recency', 'Frequency', 'Monetary']

# Display summary statistics
rfm.describe().round(2)
```

```python
from sklearn.preprocessing import StandardScaler

# Log transform to reduce extreme right-skewness
rfm_log = np.log1p(rfm[['Recency', 'Frequency', 'Monetary']])

# Standardize to zero mean and unit variance
scaler = StandardScaler()
rfm_scaled = scaler.fit_transform(rfm_log)

# Convert to DataFrame for readability
rfm_scaled_df = pd.DataFrame(rfm_scaled, columns=['Recency', 'Frequency', 'Monetary'])
rfm_scaled_df.head()
```

```python
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans

inertia = []
K_range = range(1, 9)

for k in K_range:
    km = KMeans(n_clusters=k, random_state=42, n_init=10)
    km.fit(rfm_scaled)
    inertia.append(km.inertia_)

plt.figure(figsize=(7, 4))
plt.plot(K_range, inertia, marker='o', color='#1A365D')
plt.title('Elbow Method For Optimal k')
plt.xlabel('Number of Clusters (k)')
plt.ylabel('Inertia (WCSS)')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

```python
kmeans = KMeans(n_clusters=4, random_state=42, n_init=10)
rfm['Cluster'] = kmeans.fit_predict(rfm_scaled)

# Aggregate cluster statistics
cluster_summary = rfm.groupby('Cluster').agg({
    'Recency': 'mean',
    'Frequency': 'mean',
    'Monetary': ['mean', 'count']
}).round(1)

print(cluster_summary)
```

```python
# Check which cluster ID maps to which behavior:
# - Low Recency, High Frequency, High Monetary = "Champions"
# - High Recency, Low Frequency, Low Monetary  = "At-Risk / Inactive"
# - Moderate Recency, Moderate Frequency/Spend = "Loyal / Steady"
# - Low Recency, Low Frequency, Moderate Spend = "Promising / Recent"

# Save the final segmented dataset to CSV for Power BI / GitHub
rfm.to_csv("rfm_customer_segments.csv", index=False)
print("File exported successfully!")
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

fig, axes = plt.subplots(1, 3, figsize=(15, 4))

# Recency distribution per cluster
sns.boxplot(x='Cluster', y='Recency', data=rfm, ax=axes[0], hue='Cluster', palette='Set2', legend=False)
axes[0].set_title('Recency by Cluster (Lower is Better)')

# Frequency distribution per cluster (log scale for visual clarity)
sns.boxplot(x='Cluster', y='Frequency', data=rfm, ax=axes[1], hue='Cluster', palette='Set2', legend=False)
axes[1].set_yscale('log')
axes[1].set_title('Frequency by Cluster (Higher is Better)')

# Monetary distribution per cluster (log scale)
sns.boxplot(x='Cluster', y='Monetary', data=rfm, ax=axes[2], hue='Cluster', palette='Set2', legend=False)
axes[2].set_yscale('log')
axes[2].set_title('Monetary Value by Cluster (Higher is Better)')

plt.tight_layout()
plt.savefig('cluster_distribution.png', dpi=300)
plt.show()
```
